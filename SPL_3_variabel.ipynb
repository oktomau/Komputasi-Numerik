PROGRAM PENYELESAIAN SISTEM PERSAMAAN LINEAR
MENGGUNAKAN METODE ELIMINASI GAUSS

Mata Kuliah: Komputasi Numerik
Bahasa Pemrograman: Python
Metode: Eliminasi Gauss dan Substitusi Balik
Jumlah Variabel: 3

PENJELASAN:

Bagian diatas merupakan identitas program yang menjelaskan
judul, mata kuliah, bahasa pemrograman, metode penyelesaian,
dan jumlah variabel yang digunakan.

Program ini bertujuan menyelesaikan Sistem Persamaan Linear
(SPL) tiga variabel menggunakan metode Eliminasi Gauss
dan Substitusi Balik.

```python
A = [
    [2, 1, 1],
    [1, 3, 2],
    [3, 2, 4]
]

B = [10, 18, 26]

print("Koefisien persamaan berhasil dimasukkan.")
print("Matriks A:")
for baris in A:
    print(baris)

print("Konstanta B:", B)
```

PENJELASAN:

Bagian diatas digunakan untuk memasukkan koefisien dari
tiga persamaan linear ke dalam matriks A.

Variabel A menyimpan koefisien x, y, dan z.
Variabel B menyimpan nilai konstanta dari setiap persamaan.

Perintah print() digunakan untuk menampilkan data yang
telah dimasukkan agar pengguna dapat memeriksa input.

```python
matriks = [
    A[i] + [B[i]]
    for i in range(3)
]

print("Matriks Augmented:")
for baris in matriks:
    print(baris)
```

PENJELASAN:

Bagian diatas menggabungkan matriks koefisien A dengan
vektor konstanta B menjadi matriks augmented.

Penggabungan dilakukan menggunakan operator +.
Perulangan digunakan untuk menggabungkan setiap baris.

Hasilnya adalah matriks berukuran 3 baris dan 4 kolom
yang akan digunakan dalam proses Eliminasi Gauss.

```python
# Proses eliminasi Gauss untuk membentuk matriks segitiga atas

# Membuat salinan matriks agar data awal tetap tersimpan
M = [baris.copy() for baris in matriks]

n = len(M)

# Melakukan eliminasi pada setiap kolom
for i in range(n):

    # Memilih pivot terbesar untuk mengurangi kesalahan perhitungan
    pivot = max(range(i, n), key=lambda r: abs(M[r][i]))

    # Menukar baris jika diperlukan
    M[i], M[pivot] = M[pivot], M[i]

    # Memastikan elemen pivot tidak nol
    if abs(M[i][i]) < 1e-12:
        raise ValueError("Sistem tidak memiliki solusi tunggal.")

    # Menghilangkan koefisien di bawah elemen pivot
    for j in range(i + 1, n):
        faktor = M[j][i] / M[i][i]

        for k in range(i, n + 1):
            M[j][k] -= faktor * M[i][k]

print("Matriks setelah Eliminasi Gauss:")
for baris in M:
    print([round(nilai, 4) for nilai in baris])
```

PENJELASAN:

Bagian diatas merupakan proses utama metode Eliminasi Gauss.

Pertama, matriks augmented disalin agar data awal
tidak mengalami perubahan.

Selanjutnya, program memilih elemen pivot dan melakukan
pertukaran baris jika diperlukan.

Proses eliminasi dilakukan untuk menghilangkan koefisien
di bawah elemen pivot sehingga terbentuk matriks
segitiga atas.

Jika elemen pivot bernilai nol, program akan memeriksa
apakah sistem dapat diselesaikan dengan solusi tunggal.

```python
# Menghitung nilai variabel dengan substitusi balik

x = [0] * n

# Menghitung mulai dari baris terakhir menuju baris pertama
for i in range(n - 1, -1, -1):

    jumlah = sum(M[i][j] * x[j] for j in range(i + 1, n))

    x[i] = (M[i][n] - jumlah) / M[i][i]

print("Proses substitusi balik selesai.")
```

PENJELASAN:

Bagian diatas digunakan untuk menghitung nilai variabel
setelah proses Eliminasi Gauss selesai.

Perhitungan dimulai dari baris terakhir untuk memperoleh
nilai variabel z.

Selanjutnya, program menghitung nilai y dan x
menggunakan hasil perhitungan sebelumnya.

Nilai setiap variabel disimpan dalam sebuah list
yang bernama x.

```python
# Menampilkan hasil akhir perhitungan

nama_variabel = ["x", "y", "z"]

print("HASIL PENYELESAIAN SPL 3 VARIABEL")
print("---------------------------------")

for i in range(n):
    print(f"Nilai {nama_variabel[i]} = {x[i]:.4f}")

# Memeriksa kembali hasil dengan persamaan awal
print("\nPemeriksaan hasil:")

for i in range(n):
    hasil = sum(A[i][j] * x[j] for j in range(n))
    print(f"Persamaan {i+1}: {hasil:.4f} = {B[i]}")
```

PENJELASAN:

Bagian diatas digunakan untuk menampilkan hasil akhir
penyelesaian Sistem Persamaan Linear tiga variabel.

Program menampilkan nilai x, y, dan z dengan empat
angka di belakang koma.

Selanjutnya, dilakukan pemeriksaan dengan memasukkan
kembali nilai variabel ke dalam persamaan awal.

Jika hasil perhitungan sama dengan konstanta awal,
maka solusi yang diperoleh telah memenuhi ketiga
persamaan linear tersebut.
